# Skeleton-Based Pose Classification (BlazePose)

Classifies yoga poses from 33-keypoint BlazePose skeleton data.
Compares RandomForest, SVM (RBF), and GradientBoosting.

**Requirements:** `data/raw/blazepose_yoga82/keypoints_train.csv`

```python
import sys
from pathlib import Path
ROOT = Path('.').resolve().parent
sys.path.insert(0, str(ROOT))
```

```python
# Check data availability
csv_path = ROOT / 'data' / 'raw' / 'blazepose_yoga82' / 'keypoints_train.csv'
if csv_path.exists():
    import pandas as pd
    df = pd.read_csv(csv_path)
    print(f'Loaded {len(df)} samples, {df.shape[1]} columns')
    print('Label distribution:')
    print(df['label'].value_counts())
else:
    print('keypoints_train.csv not found')
    print('Download from: https://www.kaggle.com/datasets/rashiniyasp/blazepose-skeletons-yoga-82')
```

```python
# Feature engineering demo on one sample
import numpy as np
from ml.preprocessing.skeleton_utils import extract_skeleton_features, SKELETON_FEATURE_DIM

# Simulate a landmark array for demonstration only
rng = np.random.default_rng(0)
fake_landmarks = rng.uniform(-1, 1, (33, 3)).astype(np.float32)
features = extract_skeleton_features(fake_landmarks)
print(f'Feature vector shape: {features.shape}  (expected {SKELETON_FEATURE_DIM})')
print(f'First 10 features (normalized coords): {features[:10].round(4)}')
```

```python
# Load and display metrics (after training)
import json
metrics_path = ROOT / 'models' / 'metrics' / 'skeleton_classifier_metrics.json'
if metrics_path.exists():
    with open(metrics_path) as f:
        m = json.load(f)
    for model_name, results in m.items():
        test = results['test']
        print(f"\n{model_name}")
        print(f"  Test Accuracy : {test['accuracy']}")
        print(f"  Test F1 Score : {test['f1_score']}")
        print(f"  Data Source   : {test['data_source']}")
else:
    print('Train the skeleton classifier first')
```
